```python

import numpy as np
import pandas as pd

# A list of dictionaries: each dictionary is one product
inventory = [
    {"product": "Notebook", "category": "Stationery",
     "price": 1500, "quantity": 40},
    {"product": "Pen", "category": "Stationery",
     "price": 200, "quantity": 120},
    {"product": "Backpack", "category": "Bags",
     "price": 12000, "quantity": 15},
    {"product": "Water Bottle", "category": "Drinkware",
     "price": 2500, "quantity": 0},
    {"product": "Laptop Sleeve", "category": "Bags",
     "price": 6500, "quantity": 8},
    {"product": "Highlighter Set", "category": "Stationery",
     "price": 1800, "quantity": None},
    {"product": "Coffee Mug", "category": "Drinkware",
     "price": 3000, "quantity": 22},
    {"product": "USB Flash Drive", "category": "Electronics",
     "price": 4500, "quantity": 12},
    {"product": "Earphones", "category": "Electronics",
     "price": 7000, "quantity": 5},
    {"product": "Pen", "category": "Stationery",
     "price": 200, "quantity": 120},
    {"product": "Desk Lamp", "category": "Electronics",
     "price": 9500, "quantity": 6},
]

# A simple dictionary of prices
prices = {"Notebook": 1500, "Pen": 200,
          "Backpack": 12000, "Water Bottle": 2500}

# Daily sales for 14 days
daily_sales = np.array([220, 250, 240, 230, 260, 300, 310,
                        235, 245, 255, 240, 265, 980, 320])
```

```python

# Task 1.1: Create a product dictionary

product = {
    "name": "Notebook",
    "category": "Stationery",
    "price": 1500,
    "quantity": 20
}

print(product["price"])
```

```python
# Task 1.2: Add supplier and update quantity

product["supplier"] = "ABC Suppliers"
product["quantity"] = 35

print(product)
```

```python
# Task 1.3: Loop through prices and look up a missing product

for name, price in prices.items():
    print(f"{name} costs N{price:,}.")

calculator_price = prices.get("Calculator", "Not stocked")
print(calculator_price)
```

```python
# Task 1.4: Create discounted prices using dictionary comprehension

sale_prices = {
    name: price * 0.90
    for name, price in prices.items()
}

print(sale_prices)
```

```python
# Task 1.5: Count products in each category

category_counts = {}

for item in inventory:
    category = item["category"]
    category_counts[category] = category_counts.get(category, 0) + 1

print(category_counts)
```

```python
# Task 2.1: Calculate total product value

def total_value(price, quantity):
    """Calculate the total value of a product."""
    return price * quantity

notebook_total = total_value(1500, 40)
print(notebook_total)
```

```python
# Task 2.2: Calculate discounted price

def apply_discount(price, percent=10):
    """Calculate the price after applying a discount."""
    discount = price * (percent / 100)
    return price - discount

print(apply_discount(1500))
print(apply_discount(1500, 25))
```

```python
# Task 2.3: Check the stock status of a product

def stock_status(quantity):
    """Return the stock status based on the available quantity."""
    if quantity == 0:
        return "Out of stock"
    elif quantity < 10:
        return "Low stock"
    else:
        return "In stock"

print(stock_status(0))
print(stock_status(5))
print(stock_status(40))
```

```python
# Task 2.4: Find products belonging to a category

def products_in_category(inventory, category):
    """Return product names belonging to the specified category."""
    result = []

    for item in inventory:
        if item["category"] == category:
            result.append(item["product"])

    return result

electronics = products_in_category(inventory, "Electronics")
print(electronics)
```

```python
# Task 3.1: Display the shape and data type

print("Shape:", daily_sales.shape)
print("Data type:", daily_sales.dtype)
```

```python
# Task 3.2: Calculate basic sales statistics

print("Total sales:", np.sum(daily_sales))
print("Mean sales:", round(np.mean(daily_sales), 2))
print("Highest sales:", np.max(daily_sales))
print("Standard deviation:", round(np.std(daily_sales), 2))
```

```python
# Task 3.3: Reshape daily sales into two weeks

weekly_sales = daily_sales.reshape(2, 7)

print(weekly_sales)
print("New shape:", weekly_sales.shape)
```

```python
# Task 3.4: Calculate the median and compare it with the mean

median_sales = np.median(daily_sales)
mean_sales = np.mean(daily_sales)

print("Median:", median_sales)
print("Mean:", round(mean_sales, 2))
```
